```python
# ==============================
# CHAT MODERATION SYSTEM - FULLY CORRECTED & UNIFIED (with WARNING for mild rudeness)
# ==============================
# This version:
#  - Loads data once
#  - Uses one consistent toxicity function
#  - Uses one unified user classification
#  - Uses one stable dashboard (2x2 with display_id)
#  - Works for both text and audio input
#  - Maintains history with clear/show commands
#  - No screen-jumping, no duplicate logic
#  - Adds WARNING action for mild rudeness (low toxicity, negative, high anger)

# ------------------------------
# 1. INSTALL & IMPORTS
# ------------------------------
# Uncomment and run if packages are not installed:
# !pip install transformers torch datasets scikit-learn speechrecognition gtts matplotlib pandas numpy -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import speech_recognition as sr
from gtts import gTTS
import os
import time

from datasets import load_dataset
from sklearn.metrics import roc_curve, auc, f1_score
from transformers import pipeline
from IPython.display import display, HTML, Audio, clear_output

# ------------------------------
# 2. LOAD DATASET (ONCE)
# ------------------------------
dataset = load_dataset("go_emotions")
df = dataset["train"].to_pandas()
df = df.sample(n=5000, random_state=42).reset_index(drop=True)

emotion_names = dataset["train"].features["labels"].feature.names

def get_emotion(label_list):
    return emotion_names[label_list[0]] if len(label_list) > 0 else "neutral"

df["text"] = df["text"].astype(str)
df["emotion"] = df["labels"].apply(get_emotion)

emotion_to_sentiment = {
    "joy": "POSITIVE",
    "love": "POSITIVE",
    "excitement": "POSITIVE",
    "gratitude": "POSITIVE",
    "amusement": "POSITIVE",
    "anger": "NEGATIVE",
    "sadness": "NEGATIVE",
    "fear": "NEGATIVE",
    "disgust": "NEGATIVE",
    "annoyance": "NEGATIVE",
    "neutral": "NEUTRAL"
}
df["sentiment"] = df["emotion"].map(emotion_to_sentiment).fillna("NEUTRAL")

# ------------------------------
# 3. LOAD MODELS (ONCE)
# ------------------------------
emotion_model = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=1
)

sentiment_model = pipeline(
    "text-classification",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest"
)

tox_model = pipeline(
    "text-classification",
    model="unitary/toxic-bert",
    top_k=None
)

# ------------------------------
# 4. TOXICITY FUNCTIONS (UNIFIED)
# ------------------------------
def get_toxic_prob(text):
    """
    Returns a float toxicity score (0–1).
    Uses max(prob_toxic, prob_insult, prob_threat).
    """
    result = tox_model(text)
    
    # Normalize result to a single dict
    if isinstance(result[0], list):
        result = result[0][0]
    else:
        result = result[0]
    
    if isinstance(result, list):
        d = {x["label"]: x["score"] for x in result}
    else:
        d = {result["label"]: result["score"]}
    
    return float(max(d.get("toxic", 0), d.get("insult", 0), d.get("threat", 0)))

toxic_emotions = ["anger", "disgust", "annoyance", "fear"]

df["toxicity_label"] = df["emotion"].apply(
    lambda x: 1 if x in toxic_emotions else 0
)
y_true = df["toxicity_label"].values
scores = np.array([get_toxic_prob(t) for t in df["text"].tolist()])

# ------------------------------
# 5. FIND BEST THRESHOLD (ADAPTIVE)
# ------------------------------
def find_best_threshold_adaptive(y_true, scores, target_min=0.35):
    fpr, tpr, thresholds = roc_curve(y_true, scores, pos_label=1)
    j_scores = tpr - fpr
    
    f1_scores = np.array([
        f1_score(y_true, (scores >= th).astype(int), zero_division=0) 
        for th in thresholds
    ])
    
    combined = j_scores + 0.7 * f1_scores - 0.4 * (thresholds < 0.15).astype(float)
    best_idx = np.argmax(combined)
    best_th = thresholds[best_idx]

    if best_th < target_min:
        toxic_scores = scores[y_true == 1]
        if len(toxic_scores) > 0:
            percentile_th = np.percentile(toxic_scores, 60)
            best_th = max(best_th, percentile_th)

    best_auc = auc(fpr, tpr)
    
    print(f" Final Best Threshold : {best_th:.4f}")
    print(f" AUC Score            : {best_auc:.4f}")
    print(f"F1 Score at best th   : {f1_scores[best_idx]:.4f}")
    print(f"Toxic Samples         : {y_true.sum()} / {len(y_true)}")
    
    return best_th

print("Calculating optimal toxicity threshold...")
best_th = find_best_threshold_adaptive(y_true, scores, target_min=0.35)
print(f"\n Recommended Threshold: {best_th:.4f}\n")

# ------------------------------
# 6. ROC CURVE PLOT
# ------------------------------
fpr, tpr, thresholds = roc_curve(y_true, scores, pos_label=1)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', lw=2.5, label=f'ROC Curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], [0, 1], color='gray', lw=2, linestyle='--', label='Random Guess (AUC = 0.50)')

best_idx = np.argmin(np.abs(thresholds - best_th))
plt.scatter(fpr[best_idx], tpr[best_idx], color='red', s=150, 
            marker='o', label=f'Best Threshold = {best_th:.4f}', zorder=5)

plt.annotate(
    f'Best Threshold\n{best_th:.4f}',
    xy=(fpr[best_idx], tpr[best_idx]),
    xytext=(fpr[best_idx] + 0.08, tpr[best_idx] - 0.15),
    arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
    fontsize=11, ha='center'
)

plt.xlabel("False Positive Rate (FPR)", fontsize=12)
plt.ylabel("True Positive Rate (TPR)", fontsize=12)
plt.title("ROC Curve - Toxicity Detection (Toxic-BERT)", fontsize=14)
plt.legend(loc="lower right", fontsize=11)
plt.grid(True, alpha=0.3)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.tight_layout()
plt.show()

# ------------------------------
# 7. INPUT: TEXT OR AUDIO
# ------------------------------
choice = input("Choose Input (1=Text, 2=Audio): ").strip()

def speech_to_text():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Speak now...")
        r.adjust_for_ambient_noise(source, duration=0.5)
        audio = r.listen(source, timeout=5)
    try:
        return r.recognize_google(audio)
    except Exception:
        return ""

if choice == "1":
    user_text = input("Enter text: ").strip()
elif choice == "2":
    user_text = speech_to_text()
    if not user_text:
        print("Speech not recognized. Using empty input.")
        user_text = ""
else:
    print("Invalid choice. Defaulting to text input.")
    user_text = input("Enter text: ").strip()

print("\nINPUT:", user_text)

# ------------------------------
# 8. EMOTION, SENTIMENT, TOXICITY ANALYSIS
# ------------------------------
emo = emotion_model(user_text)[0][0]
sent = sentiment_model(user_text)[0]

emotion = emo["label"].lower()
emotion_score_val = emo["score"] * 100  # 0–100

sentiment = sent["label"].upper()
sentiment_score_val = sent["score"] * 100  # 0–100

tox_score = get_toxic_prob(user_text)  # always a float

# ------------------------------
# 9. UNIFIED USER CLASSIFICATION + ACTION (with WARNING for mild rudeness)
# ------------------------------

# Helper: simple threat/slur detection (adjust to your needs)
def contains_threats_or_slurs(text: str) -> bool:
    text_lower = text.lower()
    threats = [
        "threaten", "threat", "kill you", "i will kill", "i'll kill",
        "hurt you", "i'll hurt", "i'm going to kill", "i will hurt"
    ]
    slurs = [
        "stupid", "idiot", "moron", "bastard", "bitch", "damn", "shit",
        "dumb", "retard", "faggot", "nigger"
    ]
    for word in threats + slurs:
        if word in text_lower:
            return True
    return False

# 1) Determine ACTION: WARNING, ACTIVE, or BLOCKED
#    - WARNING: mild rudeness (low toxicity, negative, moderate–high anger, no threats/slurs)
#    - BLOCKED: high toxicity or threats/slurs
#    - ACTIVE: everything else (safe)
action = "ACTIVE"  # default

# WARNING for mild rudeness
if (
    tox_score < 0.05
    and sentiment == "NEGATIVE"
    and 30 <= emotion_score_val <= 95
    and emotion == "anger"
    and not contains_threats_or_slurs(user_text)
):
    action = "WARNING"

# BLOCKED for toxic / threats / slurs
if tox_score >= best_th or contains_threats_or_slurs(user_text):
    action = "BLOCKED"

# 2) Determine USER TYPE based on toxicity + action
if action == "BLOCKED":
    user_type = "TOXIC USER"
    chat_status = "BLOCKED"
    block_time_seconds = int(tox_score * 600)
elif action == "WARNING":
    user_type = "WARNING USER"
    chat_status = "WARNING"
    block_time_seconds = 300
else:
    user_type = "SAFE USER"
    chat_status = "ACTIVE"
    block_time_seconds = 0

block_time_minutes = block_time_seconds // 60
remaining_seconds = block_time_seconds % 60

# 3) Suggestion
if emotion == "anger":
    suggestion = "Try to stay calm."
elif sentiment == "NEGATIVE":
    suggestion = "Use positive tone."
else:
    suggestion = "Good communication!"

# ------------------------------
# 10. FINAL REPORT (TEXT)
# ------------------------------
print("\n===== FINAL REPORT =====")
print("Text:", user_text)
print("Emotion:", emotion, f"({emotion_score_val:.2f}%)")
print("Sentiment:", sentiment, f"({sentiment_score_val:.2f}%)")
print("Toxicity:", f"{tox_score:.4f}")
print("Action:", chat_status)
print("User Type:", user_type)
print("Suggestion:", suggestion)
print(f"Block Time: {block_time_minutes}m {remaining_seconds}s")

# ------------------------------
# 11. SIMPLE BAR CHART (OPTIONAL)
# ------------------------------
bars = plt.bar(
    ["Emotion", "Sentiment", "Toxicity"],
    [emotion_score_val, sentiment_score_val, tox_score * 100]
)
bars[2].set_color("red")
bars[0].set_color("skyblue")
bars[1].set_color("orange")
plt.title(user_type)
plt.ylabel("Score (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()

# ------------------------------
# 12. AUDIO WARNING (IF BLOCKED OR WARNING)
# ------------------------------
def speak_and_show(text):
    filename = f"warning_{int(time.time())}.mp3"
    tts = gTTS(text)
    tts.save(filename)
    display(Audio(filename, autoplay=True))
    return filename

if chat_status == "BLOCKED":
    warning_text = (
        f"You are blocked for {block_time_minutes} minutes due to toxic behavior. "
        f"Please follow community guidelines."
    )
    print("\nTEXT WARNING:")
    print(warning_text)
    audio_file = speak_and_show(warning_text)
    print("AUDIO FILE:", audio_file)

elif chat_status == "WARNING":
    warning_text = (
        "WARNING DETECTED\n\n"
        "Your message shows inappropriate tone.\n\n"
        "Suggestion:\n"
        "- Rephrase your message politely\n"
        "- Avoid offensive language"
    )
    print("\nWARNING:")
    print(warning_text)

# ------------------------------
# 13. CHAT STATUS BAR CHART
# ------------------------------
status_map = {"ACTIVE": 1, "WARNING": 2, "BLOCKED": 3}
status_value = status_map[chat_status]
color_map = {"ACTIVE": "green", "WARNING": "orange", "BLOCKED": "red"}
color = color_map[chat_status]

plt.figure(figsize=(4, 4))
plt.bar(["Chat Status"], [status_value], color=color)
plt.text(
    0,
    status_value / 2,
    chat_status,
    ha='center',
    color='white',
    fontsize=12
)
plt.yticks([1, 2, 3], ["ACTIVE", "WARNING", "BLOCKED"])
plt.title("User Chat Status Dashboard")
plt.tight_layout()
plt.show()

# ------------------------------
# 14. HISTORY MANAGEMENT
# ------------------------------
if 'tox_history' not in globals():
    tox_history = []
if 'history_scores' not in globals():
    history_scores = []

tox_history.append({
    "Message": user_text[:55] + "..." if len(user_text) > 55 else user_text,
    "Toxicity Score": round(tox_score, 4),
    "Threshold": round(best_th, 4),
    "User Type": user_type,
    "Chat Status": chat_status,
    "Block Time": f"{block_time_minutes}m {remaining_seconds}s",
    "Emotion": emotion,
    "Sentiment": sentiment
})

history_scores.append(tox_score)
history_df = pd.DataFrame(tox_history)

print("\n" + "="*90)
print("                    CHAT MODERATION HISTORY")
print("="*90)
display(history_df)

print("\n" + "="*90)
print("CURRENT MESSAGE SUMMARY")
print("="*90)
print(f"Text           : {user_text}")
print(f"Toxicity Score : {tox_score:.4f} (Threshold: {best_th:.4f})")
print(f"Status         : {chat_status}")
print(f"User Type      : {user_type}")
print(f"Block Time     : {block_time_minutes} minutes {remaining_seconds} seconds")
print("="*90)

print("\n" + "-"*90)
print("Commands:")
print("   Type 'clear' --> Clear History")
print("   Type 'show'  --> Show History Again")
print("   Type 'skip'  --> Continue")
print("-"*90)

def clear_history():
    global tox_history, history_scores
    tox_history = []
    history_scores = []
    print("History cleared.")

command = input("Enter command (clear / show / skip): ").strip().lower()

if command == "clear":
    clear_history()
elif command == "show":
    display(pd.DataFrame(tox_history))
elif command != "skip":
    print("Unknown command. Skipping.")

# ------------------------------
# 15. STABLE 2x2 DASHBOARD (NO SCREEN JUMP)
# ------------------------------
dashboard_handle = None

def generate_moderation_dashboard(tox_score, best_th, status, user_type, block_minutes, emotion, sentiment, history_scores=None):
    global dashboard_handle
    
    status_str = str(status).strip().upper()
    user_type_str = str(user_type).strip().upper()
    is_safe = status_str in ["ACTIVE", "SAFE", "ALLOWED"] or "SAFE" in user_type_str
    
    if is_safe:
        theme_color = 'green'
        display_status = status_str
        display_subtext = user_type_str
        display_detail = "Allowed to Chat"  
    else:
        theme_color = 'red'
        display_status = status_str
        display_subtext = user_type_str
        tot_secs = int(block_minutes * 60)
        mins = tot_secs // 60
        secs = tot_secs % 60
        display_detail = f"Time Remaining : {mins:02d}:{secs:02d}"

    emotion_label = str(emotion).split()[0].upper()
    sentiment_label = str(sentiment).split()[0].upper()
    
    tox_score_pct = float(tox_score) * 100
    best_th_pct = float(best_th) * 100
    
    if not history_scores:
        history_scores = [float(tox_score)]

    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle(f"CHAT MODERATION DASHBOARD - {user_type_str}", fontsize=16, fontweight='bold', color=theme_color)

    # Subplot 1: Toxicity Level
    axs[0, 0].bar(['Toxicity Score', 'Threshold'], [tox_score_pct, best_th_pct], 
                  color=[theme_color, 'gray'])
    axs[0, 0].set_title("Toxicity Level")
    axs[0, 0].set_ylabel("Score (%)")
    axs[0, 0].set_ylim(0, 100)
    axs[0, 0].grid(True, alpha=0.3)

    # Subplot 2: Chat Status
    axs[0, 1].axis('off')
    axs[0, 1].text(0.5, 0.6, f"CHAT STATUS : {display_status}", 
                   fontsize=24, fontweight='bold', color=theme_color, ha='center')
    axs[0, 1].text(0.5, 0.45, f"User Type : {display_subtext}", 
                   fontsize=16, ha='center', color='black')
    axs[0, 1].text(0.5, 0.30, display_detail, 
                   fontsize=14, ha='center', color=theme_color, 
                   fontweight='bold' if not is_safe else 'normal')

    # Subplot 3: Emotion & Sentiment
    axs[1, 0].bar([f"Emotion\n({emotion_label})", f"Sentiment\n({sentiment_label})"], 
                  [emotion_score_val, sentiment_score_val], 
                  color=['orange', 'purple'])
    axs[1, 0].set_title("Current Emotion & Sentiment Analysis")
    axs[1, 0].set_ylabel("Confidence (%)")
    axs[1, 0].set_ylim(0, 100)
    axs[1, 0].grid(True, alpha=0.3)

    # Subplot 4: Toxicity History
    axs[1, 1].plot(range(1, len(history_scores) + 1), history_scores, 
                   marker='o', color=theme_color, linewidth=3)
    axs[1, 1].axhline(y=float(best_th), color='gray', linestyle='--', label='Threshold')
    axs[1, 1].set_title("Toxicity History")
    axs[1, 1].set_xlabel("Message Number")
    axs[1, 1].set_ylabel("Toxicity Score (0-1)")
    axs[1, 1].set_ylim(0, 1.05)  
    axs[1, 1].legend()
    axs[1, 1].grid(True)

    plt.tight_layout()
    
    if dashboard_handle is None:
        dashboard_handle = display(fig, display_id=True)
    else:
        dashboard_handle.update(fig)
        
    plt.close(fig)

# Generate dashboard
generate_moderation_dashboard(
    tox_score=tox_score,
    best_th=best_th,
    status=chat_status,
    user_type=user_type,
    block_minutes=block_time_minutes + (remaining_seconds / 60.0),
    emotion=emotion,
    sentiment=sentiment,
    history_scores=history_scores
)

# ------------------------------
# 16. OPTIONAL: COUNTDOWN TIMER FOR BLOCKED USERS
# ------------------------------
def chat_dashboard_with_timer(minutes, user_type):
    if user_type != "TOXIC USER":
        print(f"Timer not required. User type '{user_type}' is not blocked.")
        return

    total_seconds = int(minutes * 60)

    for remaining in range(total_seconds, -1, -1):
        mins = remaining // 60
        secs = remaining % 60

        clear_output(wait=True)

        print("=" * 40)
        print("        CHAT MODERATION DASHBOARD        ")
        print("=" * 40)
        print(f"User Type      : {user_type}")
        print("\nCHAT STATUS    :  BLOCKED")
        print(f"Time Remaining : {mins:02d}:{secs:02d}")
        print("=" * 40)
        print("Please wait... System monitoring active.")

        if remaining > 0:
            time.sleep(1)

    clear_output(wait=True)
    print(" CHAT UNBLOCKED - USER CAN CHAT NOW ")

# Uncomment to enable countdown timer for blocked users:
# chat_dashboard_with_timer(block_time_minutes, user_type)
```
